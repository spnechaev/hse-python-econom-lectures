## Когда нужен класс

Класс связывает состояние объекта с операциями, которые должны сохранять правила этого состояния. Например, у банковского счёта есть баланс, а операции пополнения и списания обязаны проверять валюту, доступный остаток и ограничения счёта. Если передать баланс отдельно от этих операций, правила легко обойти. Здесь класс удерживает данные и проверки вместе.

Если данные и функции живут независимо, класс не нужен. Оборачивать каждую пару функций в `Manager`, `Service` или `Helper` — не ООП, а лишний уровень имён.

Основные идеи:

- **Абстракция:** внешний код работает с понятным контрактом и не зависит от деталей реализации.
- **Инкапсуляция:** операции объекта не дают привести его состояние к недопустимому виду; внутренние детали без необходимости не входят в публичный контракт.
- **Наследование:** подкласс уточняет контракт базового класса и может использоваться вместо него. Это отношение «является», а не универсальный способ сократить код.
- **Полиморфизм:** один и тот же клиентский код работает с разными объектами через общий контракт. В Python для этого часто достаточно duck typing; наследование не обязательно.

## Экземпляр, класс и метод

Экземпляр хранит собственное состояние, а класс описывает общее поведение экземпляров и может хранить общие атрибуты. При вызове `pet.say()` Python находит функцию `say` в классе и передаёт объект `pet` первым аргументом. Имя `self` не является ключевым словом, но отступать от этого соглашения нельзя без веской причины: другой разработчик ожидает увидеть экземпляр именно под этим именем.

Состояние лучше менять через операции предметной области, которые проверяют инварианты — правила, остающиеся истинными после любой допустимой операции. Для счёта таким правилом может быть запрет отрицательного остатка, для животного в примере ниже — непустое имя.

Механические пары `get_x()`/`set_x()` для каждого поля ничего не инкапсулируют. Если при обычном присваивании нужна проверка, в Python используют `property`.


````python
class Animal:
    def __init__(self, name: str) -> None:
        self.name = name

    @property
    def name(self) -> str:
        return self._name

    @name.setter
    def name(self, value: str) -> None:
        value = value.strip()
        if not value:
            raise ValueError("Name must not be empty")
        self._name = value

    def __str__(self) -> str:
        return f"Animal named {self.name}"

    def say(self) -> None:
        print("Animal sound is not defined")

pet = Animal("Lucky")
pet2 = Animal("Molly")
print(pet)
pet.say()
pet.name = "Marquise"
print(pet)
pet2.say()
````

## Атрибуты экземпляра и класса

При чтении `obj.attribute` Python сначала ищет атрибут у экземпляра, затем в его классе и базовых классах. Присваивание `obj.legs = 3` создаёт атрибут именно у этого экземпляра. Он заслоняет одноимённый атрибут класса, но не меняет ни класс, ни другие экземпляры.

Изменяемый объект вроде списка в атрибуте класса будет общим для всех экземпляров. Это не набор личных записных книжек, а одна общая доска: изменение, сделанное через один экземпляр, увидят остальные. Индивидуальные списки и словари нужно создавать в `__init__`.

- Обычный метод получает экземпляр `self`.
- `classmethod` получает класс `cls`; его часто используют для альтернативных конструкторов, которые создают объект из другого формата данных.
- `staticmethod` не получает ни экземпляр, ни класс. Это обычная функция, помещённая в пространство имён класса; если она не относится к классу по смыслу, ей место на уровне модуля.


````python
class Animal:
    legs: int = 4

    def __init__(self, name: str) -> None:
        if not self.is_valid_name(name):
            raise ValueError("Name must not be empty")
        self.name = name

    def __str__(self) -> str:
        return f"Animal named {self.name}, legs: {self.legs}"

    @classmethod
    def from_description(cls, description: str) -> "Animal":
        _, name = description.split(":", maxsplit=1)
        return cls(name.strip())

    @staticmethod
    def is_valid_name(name: str) -> bool:
        return bool(name.strip())


class Dog(Animal):
    pass


pet = Animal("Lucky")
another_pet = Animal.from_description("name: Marquise")
dog = Dog.from_description("name: Rex")
assert isinstance(dog, Dog)

# Атрибут экземпляра заслоняет атрибут Animal.legs только для pet.
pet.legs = 3
print(pet)
print(another_pet)

# Изменение класса видно экземплярам без собственного атрибута legs.
Animal.legs = 5
print(another_pet)
print(pet)
````

## Наследование и полиморфизм

Наследование уместно, когда объект подкласса действительно выполняет контракт базового класса и может использоваться вместо него. Подкласс вправе переопределять и добавлять методы, но смысл унаследованных операций должен сохраняться. Если код ожидает `Animal`, передача `Dog` не должна требовать дополнительных проверок и особых обходных путей. Иерархия, построенная только ради повторного использования пары методов, обычно хуже композиции.

Полиморфизм виден в коде, который вызывает один и тот же метод у разных объектов и не содержит цепочку `if type(...)`. Циклу ниже не важно, лежит перед ним собака или кошка: достаточно метода `talk()`. Если для каждого нового животного приходится добавлять отдельную ветку, общего контракта фактически нет.

## Соглашения о видимости

- `name` — публичная часть интерфейса.
- `_name` — внутренний атрибут по соглашению. Доступ технически открыт, но внешний код не должен на него опираться.
- `__name` запускает name mangling: внутри `Animal` имя станет `_Animal__name`. Это защищает прежде всего от случайного конфликта имён в наследнике, а не от доступа. Настоящего модификатора private в Python нет.

Одинарное подчёркивание похоже на табличку «Служебное помещение»: дверь технически открывается, но внешний код не должен считать находящиеся за ней детали частью доступного сервиса. Если опереться на внутренний атрибут, следующий рефакторинг закономерно сломает такой код.

## Композиция

Наследование выражает отношение «является», композиция — «содержит» или «использует». Автомобиль использует двигатель, но не является двигателем. Двигатель можно заменить, не превращая автомобиль в другой вид автомобиля. Если часть поведения должна меняться независимо от остального объекта, композиция обычно точнее описывает модель:

```python
class Engine:
    def start(self) -> None:
        print("Engine started")


class Car:
    def __init__(self, engine: Engine) -> None:
        self._engine = engine

    def start(self) -> None:
        self._engine.start()
```

## Множественное наследование и MRO

При нескольких базовых классах Python заранее строит порядок разрешения атрибутов — MRO. Его можно увидеть в `Class.__mro__`. Поиск метода идёт по этому списку слева направо, а `super()` продолжает поиск со следующей позиции. Поэтому `super()` ведёт не к абстрактному «родителю сверху», а к конкретному следующему классу в MRO.

Множественное наследование узких интерфейсов и небольших mixin-классов может быть оправдано. Несколько базовых классов с собственным состоянием, несовместимыми `__init__` и одноимёнными методами быстро делают порядок вызовов неочевидным. В такой ситуации композиция обычно проще.


````python
from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def talk(self) -> None:
        ...


class Dog(Animal):
    def talk(self) -> None:
        print("woof")


class Cat(Animal):
    def talk(self) -> None:
        print("meow")


class LoudCat(Cat):
    def talk(self) -> None:
        print("hiss")
        super().talk()


animals = [Dog(), Cat(), LoudCat()]
for animal in animals:
    animal.talk()


# Dog стоит раньше Cat, поэтому CatDog.talk берётся из Dog.
class CatDog(Dog, Cat):
    pass


CatDog().talk()
print([cls.__name__ for cls in CatDog.__mro__])


class Flyable(ABC):
    @abstractmethod
    def fly(self) -> None:
        ...
    

class Swimmable(ABC):
    @abstractmethod
    def swim(self) -> None:
        ...


class BaseFlyer(Flyable):
    def fly(self) -> None:
        print("I'm flying")


class FlyingFish(BaseFlyer, Swimmable):
    def swim(self) -> None:
        print("I'm swimming deep")


class Duck(Flyable, Swimmable):
    def fly(self) -> None:
        print("I'm flying")

    def swim(self) -> None:
        print("I'm swimming")


fish = FlyingFish()
fish.fly()
fish.swim()
````

## Итоги

- Класс полезен, когда состояние и операции образуют единый объект с инвариантами.
- Атрибуты экземпляра индивидуальны, атрибуты класса переиспользуются, пока экземпляр их не заслонит.
- `property` позволяет сохранить обычный синтаксис атрибута и контролировать присваивание.
- Наследование обязано сохранять контракт базового класса; повторное использование кода само по себе его не оправдывает.
- Полиморфизм убирает проверки конкретных типов из клиентского кода.
- При множественном наследовании поиск и `super()` следуют MRO.
- Композиция подходит для частей, которые объект содержит, использует или должен независимо заменять.
