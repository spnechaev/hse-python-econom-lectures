## Программа и алгоритм

Алгоритм описывает способ решения задачи, а программа переводит этот способ в команды для конкретной среды. Примерно так же рецепт описывает, что нужно сделать с продуктами, а на конкретной кухне приходится учитывать плиту, посуду и доступные инструменты.

Один алгоритм можно реализовать на разных языках. В программе к нему обычно добавляются ввод и вывод данных, проверки, обработка ошибок и взаимодействие с другими частями системы.


## Что требуется от алгоритма

**Алгоритм** — конечная последовательность однозначных шагов, которая для допустимых входных данных должна дать требуемый результат. Слова «допустимые» и «требуемый» нельзя оставлять на усмотрение читателя. До реализации нужно определить:

- допустимые входные данные;
- ожидаемый результат;
- поведение на граничных случаях;
- условие завершения.

Алгоритм можно описать словами, псевдокодом, блок-схемой или формулами. Программа требует большей точности: нужно выбрать типы данных, договориться о граничных случаях и использовать возможности конкретной среды выполнения.

Делить сами языки на строго «интерпретируемые» и «компилируемые» неправильно: это характеристика реализации, а не языка как такового. Например, CPython сначала компилирует исходный код в байткод, а затем выполняет байткод на виртуальной машине. Другие реализации Python могут устроить выполнение иначе.


## Пример: поиск максимального элемента

Пусть на вход подаётся список целых чисел. Для пустого списка возвращаем `None`. В непустом списке действуем так:

1. Запоминаем первый элемент как текущий максимум.
2. Просматриваем остальные элементы слева направо.
3. Если очередной элемент больше текущего максимума, заменяем максимум.
4. После обхода возвращаем найденное значение.

Это тот же способ, которым ищут самого высокого человека в ряду: сначала считают первым кандидатом первого человека, затем сравнивают с ним каждого следующего. Пропустить кого-то нельзя — именно он может оказаться выше всех.

Описание уже достаточно точное, чтобы человек мог выполнить его вручную. Теперь запишем те же шаги на Python.


````python
def find_max(items: list[int]) -> int | None:
    if not items:
        return None

    current_max = items[0]
    for index in range(1, len(items)):
        item = items[index]
        if item > current_max:
            current_max = item
    return current_max


print(find_max([1, 2, 3, 4, 5]))
````

## Характеристики алгоритма

Оценка сложности отвечает не на вопрос «сколько секунд займёт программа», а на вопрос «как изменятся затраты, если вход станет больше». На кассе один сотрудник может сканировать товары быстрее другого, но удвоение числа товаров примерно удвоит объём работы для любого из них. Нас интересует эта зависимость, а не скорость конкретной кассы.

Сначала определяют размер входа. Для списка естественно взять `N = len(items)`. Для графа одного числа может быть мало: отдельно учитывают количество вершин и рёбер.

**Временная сложность** оценивает количество выбранных элементарных операций. **Пространственная сложность** оценивает расход памяти; обычно отдельно считают дополнительную память, которая нужна алгоритму помимо хранения входных данных.

`T(N) = O(f(N))` означает, что начиная с некоторого размера входа значение `T(N)` не превосходит константу, умноженную на `f(N)`. Big O задаёт верхнюю границу, а не точную оценку. Линейная функция формально принадлежит и `O(N²)`, но это всё равно что оценить десятиминутную работу фразой «займёт не больше суток»: утверждение верное, пользы от него мало. Для точного порядка роста используют `Θ`: поиск максимума работает за `Θ(N)`.

| Обозначение | Смысл |
|---|---|
| `O(f(N))` | асимптотическая верхняя граница |
| `Ω(f(N))` | асимптотическая нижняя граница |
| `Θ(f(N))` | верхняя и нижняя границы совпадают по порядку роста |

При асимптотической оценке отбрасывают постоянные множители и младшие слагаемые: `3N + 10 = Θ(N)`, а `N² + N = Θ(N²)`. Мы сравниваем характер роста при больших `N`, но не утверждаем, что константы не важны. На реальных данных два алгоритма с одинаковой асимптотикой могут работать с заметно разной скоростью.

## Лучший, средний и худший случаи

В худшем случае считают максимальные затраты среди всех допустимых входов размера `N`, в лучшем — минимальные. Для среднего случая нужно задать вероятностное распределение входов. Фраза «в среднем быстро» без такой модели ничего не доказывает.

У поиска максимума в неупорядоченном списке нет короткого пути: нужно проверить каждый элемент. Как и в ряду людей, любой пропущенный участник может оказаться самым высоким. Алгоритм выполняет `N - 1` сравнений, поэтому время — `Θ(N)`, а дополнительная память — `Θ(1)`.

Нужно учитывать и скрытую работу Python. Срез `items[1:]` не просто обозначает остаток списка, а создаёт новый список и копирует в него ссылки на элементы. Это требует `Θ(N)` времени и памяти, поэтому выше используется обход по индексам. Встроенный `max()` тоже просматривает вход и работает за `Θ(N)`.

Если список **уже** отсортирован по возрастанию, максимум находится в последнем элементе и доступен за `Θ(1)`. Но сортировать неупорядоченный список только ради максимума бессмысленно: мы выполним примерно `O(N log N)` работы вместо одного прохода за `Θ(N)`.

## Итоги

- Алгоритм задаёт общий способ решения, программа — его исполняемую реализацию.
- Корректность требует определить допустимый вход, результат и граничные случаи.
- Асимптотика показывает рост затрат, а не точное время выполнения.
- Big O задаёт верхнюю границу; `Θ` описывает точный порядок роста.
- Пространственная оценка должна учитывать скрытые копии и временные структуры.
- Поиск максимума в неупорядоченном списке требует `Θ(N)` времени и `Θ(1)` дополнительной памяти.

**Сигнатура**

```python
def find_missing_number(items: list[int]) -> int:
    ...
```

**Примеры**

```python
find_missing_number([1, 2, 4]) == 3
find_missing_number([1, 2, 3]) == 4
find_missing_number([]) == 1
```

**Критерии проверки**

- результат корректен для пустого списка и случая, когда отсутствует `N + 1`;
- время `O(N)`, дополнительная память `O(1)`;
- входной список не изменяется.


````python
def find_missing_number(items: list[int]):
    if not items:
        return 1
    for i in range(1, len(items) + 2):  # не забываем проверить N + 1
        if i not in items:  # Поиск i в списке требует до N сравнений.
            return i

# Сложность: O(N²) по времени.
````

## Битовые операции и XOR

Битовые операции применяются отдельно к каждому разряду двоичного представления целых чисел:

| Операция | Символ | `a = 0b101` | `b = 0b011` | Результат |
|---|:---:|:---:|:---:|:---:|
| AND | `&` | `101` | `011` | `001` |
| OR | `\|` | `101` | `011` | `111` |
