## CPU-bound и I/O-bound задачи

Сначала нужно понять, на что программа тратит время. От характера нагрузки зависит подходящий способ конкурентного выполнения.

| Тип нагрузки | На что уходит основное время | Примеры | Типичный подход |
|---|---|---|---|
| CPU-bound | Вычисления на процессоре | обработка изображений, сжатие, численные расчёты | процессы, специализированные библиотеки, профилирование и оптимизация |
| I/O-bound | Ожидание сети, диска или внешнего сервиса | HTTP-запросы, драйвер БД, сокеты | `asyncio` с асинхронными библиотеками или потоки для блокирующего API |

Представьте сотрудника, который ведёт несколько заявок. Отправив запрос во внешнюю службу, он не сидит без дела до ответа, а берёт следующую заявку. Работа над несколькими заявками продвигается за один промежуток времени, хотя в каждый момент сотрудник занят только одной. Это конкурентность.

Параллелизм возникает, когда несколько сотрудников одновременно работают над разными заявками. В программе это означает физически одновременное выполнение нескольких вычислений. `asyncio` в одном потоке даёт конкурентность, а не параллелизм: пока одна задача ждёт I/O, event loop передаёт выполнение другой готовой задаче.

В обычной сборке CPython с GIL только один поток процесса исполняет Python-байткод в конкретный момент. Для `asyncio` важнее другое: задача сама возвращает управление event loop в точке ожидания. Долгий код без подходящего `await` не даёт переключиться на другие задачи и блокирует весь цикл событий.

## Корутина, задача и await

- `async def` объявляет корутинную функцию. Её вызов создаёт объект корутины, но не запускает тело функции.
- `await` ожидает awaitable-объект. Если результат ещё не готов, текущая корутина сохраняет своё состояние и возвращает управление event loop. Если результат уже готов, переключения может не произойти.
- `asyncio.Task` регистрирует корутину в event loop для конкурентного выполнения. Создать объект корутины и создать задачу — не одно и то же.
- `asyncio.TaskGroup` задаёт общую область жизни связанных задач. Выход из блока ждёт их завершения; необработанная ошибка одной задачи отменяет остальные, после чего ошибки передаются вызывающему коду.

## Как работает event loop

```text
очередь готовых задач ──► event loop ──► выполняет задачу до следующего await
        ▲                                      │
        │                                      ▼
        └──── завершение I/O ◄──── ОС / сеть / таймер
```

В упрощённом виде цикл работает так:

1. Event loop выбирает задачу, готовую продолжить работу.
2. Задача выполняется до завершения или до операции, результат которой ещё не готов.
3. Ожидание регистрируется в соответствующем механизме ОС, а event loop выбирает следующую готовую задачу.
4. Когда I/O, таймер или другое ожидание завершается, исходная задача снова попадает в очередь готовых.

Само наличие `async def` или `await` не превращает синхронную операцию в неблокирующую. `time.sleep()`, обычный файловый API или тяжёлый цикл внутри корутины по-прежнему занимают поток и останавливают event loop. Блокирующую I/O-функцию при необходимости можно выполнить в рабочем потоке:

```python
result = await asyncio.to_thread(blocking_function, argument)
```

Для CPU-bound вычислений `to_thread()` в обычной сборке CPython обычно не обеспечивает параллельного выполнения Python-кода из-за GIL. Для таких вычислений чаще используют процессы или библиотеку, которая выполняет работу вне GIL.

## Ошибки, отмена и тайм-ауты

Внешняя операция может не ответить вообще, поэтому ожидание должно иметь границу. В Python 3.11+ произвольный асинхронный блок можно ограничить по времени через `asyncio.timeout()`:

```python
async with asyncio.timeout(5):
    result = await slow_operation()
```

При отмене задача получает `asyncio.CancelledError` в очередной точке приостановки. Соединения, файлы и другие ресурсы освобождают через контекстные менеджеры или `try/finally`. Если `CancelledError` перехвачен ради очистки, после очистки его обычно нужно передать дальше; подавленная отмена нарушает управление жизненным циклом задач.


## Первая корутина

В Jupyter event loop уже работает, поэтому корутину можно ожидать через `await` прямо в ячейке. В обычном `.py`-файле цикл событий обычно создаёт и закрывает `asyncio.run(main())`.


````python
import asyncio


async def process_request(request: str) -> str:
    print(f"Started: {request}")
    await asyncio.sleep(0.1)
    return f"Processed {request}"


result = await process_request("aaa")
print(result)
````

## Конкурентные HTTP-запросы

Один `ClientSession` переиспользует пул соединений, поэтому его создают на группу запросов, а не заново для каждого URL. `TaskGroup` запускает запросы конкурентно и дожидается всех созданных в нём задач. В примере ожидаемые сетевые ошибки перехватываются внутри `check()` и становятся результатом конкретной проверки, поэтому недоступность одного адреса не отменяет остальные.


````python
import asyncio
import aiohttp


async def check(
    session: aiohttp.ClientSession, url: str
) -> tuple[str, int | str]:
    try:
        async with session.get(url) as response:
            response.raise_for_status()
            return url, response.status
    except (aiohttp.ClientError, TimeoutError) as error:
        return url, f"{type(error).__name__}: {error}"


async def check_all(websites: list[str]) -> list[tuple[str, int | str]]:
    timeout = aiohttp.ClientTimeout(total=5)
    async with aiohttp.ClientSession(timeout=timeout) as session:
        async with asyncio.TaskGroup() as group:
            tasks = [
                group.create_task(check(session, url))
                for url in websites
            ]
    return [task.result() for task in tasks]


websites = [
    "https://www.python.org",
    "https://docs.aiohttp.org",
    "https://example.com/not-found",
]
for url, result in await check_all(websites):
    print(f"{url}: {result}")
````

## Блокирующий и неблокирующий вызов

Обе функции объявлены через `async def`, но `time.sleep()` удерживает поток, а `asyncio.sleep()` приостанавливает текущую задачу и отдаёт управление event loop. Поэтому пять вызовов `blocking_sleep()` фактически выполняются один за другим, а ожидания в пяти вызовах `nonblocking_sleep()` перекрываются по времени.


````python
import asyncio
import time


DELAY = 0.2
CALL_COUNT = 5


async def blocking_sleep() -> None:
    time.sleep(DELAY)


async def nonblocking_sleep() -> None:
    await asyncio.sleep(DELAY)


async def measure(operation) -> float:
    start = time.perf_counter()
    async with asyncio.TaskGroup() as group:
        for _ in range(CALL_COUNT):
            group.create_task(operation())
    return time.perf_counter() - start


blocking_time = await measure(blocking_sleep)
nonblocking_time = await measure(nonblocking_sleep)
print(f"Blocking calls: {blocking_time:.2f} seconds")
print(f"Non-blocking calls: {nonblocking_time:.2f} seconds")
````

## Где пригодится в индивидуальном проекте

Асинхронность полезна, когда проект одновременно ждёт несколько независимых внешних операций: обращается к API, работает с асинхронным драйвером БД или обслуживает много сетевых соединений. Пока одна операция ждёт внешний ответ, event loop может продвигать остальные. Тяжёлые вычисления от этого быстрее не станут.

В асинхронном пути нельзя случайно оставить блокирующую библиотеку: один синхронный HTTP-запрос внутри `async def` остановит обработку остальных задач в том же event loop. Для каждого внешнего вызова нужны тайм-аут, обработка ошибок и понятная политика отмены. Число одновременных запросов иногда ограничивают семафором, чтобы не исчерпать локальные ресурсы и не перегрузить внешний сервис.

## Итоги

- `asyncio` подходит прежде всего для конкурентных I/O-bound операций.
- `await` создаёт точку кооперативного переключения только при ожидании незавершённой операции.
- Блокирующий код останавливает event loop, даже если находится внутри `async def`.
- `TaskGroup` задаёт явные границы жизни связанных задач и корректно обрабатывает их совместное завершение.
- Тайм-ауты, отмена и освобождение ресурсов — обязательная часть надёжного асинхронного кода.
