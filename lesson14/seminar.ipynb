# Семинар 14. Защита групповых проектов

Защита показывает не количество написанных файлов, а законченный пользовательский сценарий, качество решений и понимание проекта командой. Это последняя контрольная точка работы, начатой на занятии 4.

## Цели

На защите команда должна:

- кратко назвать пользователя, проблему и критерий успеха;
- показать работающий сквозной сценарий;
- объяснить архитектуру на уровне крупных частей;
- разобрать одно содержательное техническое решение;
- обосновать выбранные технологии и отказ от лишних;
- показать воспроизводимый запуск и автоматические проверки;
- честно назвать известные ограничения;
- ответить на вопросы по коду и вкладу участников.

Остальные слушатели тренируются анализировать чужие решения конструктивно и предметно.

## Перед началом

Точный порядок команд и продолжительность выступления преподаватель сообщает отдельно с учётом числа проектов. Даты и минуты в учебном notebook не фиксируются. Каждая команда заранее проверяет демонстрацию на том устройстве и в той среде, которые будут использоваться на защите.

Подготовьте репозиторий, README, команду запуска, тесты и основной набор данных. Записанное видео или скриншоты допустимы как резерв при сбое внешнего сервиса, но не заменяют работоспособный проект и ответы по коду.

## Рекомендуемая структура выступления

1. **Контекст:** кто пользователь и какую проблему решает проект.
2. **Критерий успеха:** какой наблюдаемый результат означает, что решение работает.
3. **Демонстрация:** вход, основное действие и результат без экскурсии по каждому экрану.
4. **Архитектура:** предметное ядро, внешние интерфейсы, хранение и направления зависимостей.
5. **Техническое решение:** один алгоритм, структура данных, интеграция или компромисс с альтернативами.
6. **Качество:** тесты, обработка ошибок, конфигурация и воспроизводимый запуск.
7. **Ограничения:** что сознательно не вошло и почему.
8. **Вопросы:** ответы участников по всему проекту.

Демонстрация идёт раньше архитектурного рассказа: аудитория сначала видит, зачем существует система.

## Один сквозной сценарий лучше каталога функций

Плохая демонстрация перечисляет меню, классы и библиотеки. Хорошая начинается с реального входа и заканчивается полезным результатом. Например: загрузить файл операций → увидеть обнаруженные проблемы → исправить одну запись → получить отчёт.

Дополнительные возможности показываются только если объясняют важное решение. Не тратьте основное время на регистрацию пользователя, если предметная ценность проекта находится в прогнозе, оптимизации или анализе данных.

## Как объяснять архитектуру

Покажите несколько крупных частей и связи между ними:

- откуда приходит ввод;
- где находятся предметные правила;
- где выполняются алгоритмы;
- как хранятся данные;
- какие внешние системы вызываются;
- где переводятся и обрабатываются ошибки.

Список из двадцати имён файлов не объясняет архитектуру. Достаточно схемы или дерева каталогов, если по нему видно направление зависимостей. Подготовьтесь открыть конкретный фрагмент кода по вопросу преподавателя.

## Содержательное техническое решение

Выберите один участок, где действительно было что решать:

- алгоритм и его сложность;
- структура данных и инвариант;
- схема базы и транзакция;
- конкурентная обработка и ограничение нагрузки;
- граница между предметным ядром и фреймворком;
- обработка ненадёжного внешнего API;
- способ тестирования трудно воспроизводимого сценария.

Объясните исходную проблему, рассмотренные варианты, выбранный компромисс и доказательство корректности. Фраза «взяли библиотеку, потому что она популярная» не является обоснованием.

## Что служит доказательством

| Утверждение | Наблюдаемое доказательство |
| --- | --- |
| Проект решает задачу | Сквозная демонстрация с понятным результатом |
| Запуск воспроизводим | README и запуск из чистого окружения |
| Правила работают | Автоматические тесты существенных случаев |
| Ошибки учтены | Контролируемый сбой и понятное сообщение |
| Алгоритм подходит | Инвариант, оценка сложности или измерение |
| Данные согласованы | Ограничения схемы и транзакционная граница |
| Технология оправдана | Сравнение с более простым вариантом |
| Команда понимает код | Ответы разных участников на вопросы |

Слайды сами по себе не доказывают работоспособность.

## Критерии разбора проекта

При оценивании учитываются:

- соответствие результата заявленной задаче;
- работоспособность основного сценария;
- воспроизводимость установки и запуска;
- качество предметного ядра и границ модулей;
- корректность алгоритмов, структур данных и хранения;
- тесты и явная обработка ошибок;
- обоснованность технологий;
- качество README и демонстрации;
- понимание проекта каждым участником;
- честность относительно ограничений и собственного вклада.

Количество технологий, строк кода, endpoint или таблиц не является самостоятельным критерием качества.

## Связь с итоговой оценкой

Проект — один из четырёх компонентов итоговой десятибалльной оценки наряду с семинарами, домашними работами и экзаменом. Точные коэффициенты и правила округления публикуются отдельно.

Защита позволяет оценить как общий результат команды, так и понимание отдельных участников. Одинаковое присутствие в списке авторов само по себе не гарантирует одинаковой оценки, если ответы и история работы показывают существенно разный вклад и понимание.

## Участие команды

Каждый участник должен быть готов:

- объяснить задачу и архитектуру целиком;
- назвать собственный вклад;
- открыть и разобрать связанный код;
- ответить о тестах, ошибках и ограничениях;
- объяснить решение другого участника на базовом уровне.

Разделение труда нормально: один человек мог больше заниматься интерфейсом, другой алгоритмом. Но проект не должен превращаться в несколько неизвестных друг другу частей, соединённых вечером перед защитой.

## Типовые вопросы на защите

- Почему проблема сформулирована именно так?
- Какой минимальный сценарий уже приносит пользу?
- Что произойдёт на пустом, большом или некорректном входе?
- Где находится главный инвариант?
- Какова сложность ключевой операции?
- Почему выбран этот контейнер, алгоритм или схема?
- Что произойдёт при отказе сети или базы?
- Где начинается и заканчивается транзакция?
- Какие тесты поймают наиболее дорогую ошибку?
- Почему здесь нужен async, Docker, FastAPI или ML?
- Что станет узким местом при росте нагрузки?
- Как воспроизвести результат на другой машине?
- Что вы сознательно не сделали?
- Какой следующий шаг имеет наибольшую ценность?

Вопрос не обязательно означает найденный недостаток; часто он проверяет границы понимания решения.

## План на случай технического сбоя

Перед выступлением:

- сохраните небольшой локальный набор данных;
- проверьте токены и лимиты внешних API, не показывая секреты;
- подготовьте команду полного запуска, а не историю терминала;
- закройте лишние приложения и уведомления;
- имейте короткое видео или скриншоты основного сценария;
- умейте продолжить объяснение архитектуры без внешнего сервиса.

Резерв показывает подготовку, но не должен маскировать проект, который никогда не запускался воспроизводимо.

## Что обычно мешает защите

- Долгая история идеи вместо работающего результата.
- Демонстрация через заранее подготовленную строку в базе без реального входа.
- Чтение кода построчно вместо объяснения решения.
- Архитектурная схема, не совпадающая с репозиторием.
- Фраза «за это отвечал другой участник» на каждый вопрос.
- Секреты, персональные данные или production-токены на экране.
- Обещание несуществующих возможностей в будущем времени.
- Сокрытие известного ограничения вместо объяснения компромисса.
- Перезапуск сервера и установка пакетов в середине отведённого времени без необходимости.

Спокойное признание ограничения лучше импровизированной неправды: задача защиты — показать инженерное понимание.

## Работа аудитории

Во время чужой защиты зафиксируйте:

1. какую проблему решает проект;
2. какой сценарий это доказал;
3. какое решение показалось сильным;
4. какой вопрос помог бы уточнить границу системы.

Хороший вопрос конкретен и относится к наблюдаемому решению: «Что произойдёт с двумя одновременными обновлениями заказа?» полезнее, чем «Почему не использовали другой фреймворк?». Сравнивать проекты по сложности темы без учёта качества реализации бессмысленно.

## Финальный комплект проекта

К моменту защиты в репозитории должны быть доступны:

- README с задачей, архитектурой, установкой, запуском и ограничениями;
- исходный код без секретов и локальных артефактов;
- описание зависимостей;
- автоматические тесты;
- пример конфигурации без настоящих ключей;
- небольшой демонстрационный набор данных, если он нужен;
- осмысленная история Git;
- лицензии и ссылки на внешние данные или модели, если применимо.

Точная форма передачи ссылки и фиксации версии проекта объявляется преподавателем отдельно.

## После защиты

Новой домашней работы после занятия 14 нет. Если преподаватель явно просит исправить блокирующий дефект или предоставить недостающий артефакт, команда согласует конкретный объём и способ проверки. Это не разрешение бесконечно дописывать проект после демонстрации.

Даже если разработка заканчивается, полезно зафиксировать короткую ретроспективу: что сработало, какое решение оказалось дорогим, что команда сделала бы иначе и какой следующий шаг действительно нужен пользователю.

## Самопроверка

1. Можете ли вы назвать пользователя и проблему одним предложением?
2. Какой один сценарий доказывает полезность проекта?
3. Запускается ли он по README в чистом окружении?
4. Какие крупные части есть в архитектуре и куда направлены зависимости?
5. Какое техническое решение вы разберёте глубже?
6. Какие альтернативы рассматривались?
7. Какие существенные ошибки покрывают тесты?
8. Как проект ведёт себя при отказе внешней системы?
9. Где находятся секреты и как они передаются?
10. Какой алгоритм или структура данных требует оценки сложности?
11. Как каждый участник объяснит свой вклад?
12. Какие ограничения вы назовёте сами, не дожидаясь вопроса?
13. Есть ли резервный сценарий при сетевом сбое?
14. Не занимает ли второстепенная возможность большую часть демонстрации?
15. Что останется в финальном репозитории после защиты?

## Итоги

- Защита строится вокруг пользователя, проблемы и работающего результата.
- Один связный сценарий информативнее перечня функций.
- Архитектура объясняется крупными частями и направлениями зависимостей.
- Содержательное решение включает альтернативы, компромисс и доказательство.
- Тесты, README и чистый запуск служат наблюдаемыми доказательствами качества.
- Каждый участник понимает проект целиком и способен разобрать свой вклад.
- Известные ограничения называются честно.
- FastAPI, Docker, база и async оцениваются только через пользу для задачи.
- Точный организационный регламент сообщается отдельно.
- После занятия 14 новой домашней работы нет.
