## Зачем нужна база данных
Приложению часто нужно сохранять состояние между запусками: учётные записи, историю покупок, складские остатки, результаты наблюдений или пользовательские публикации. В простом сценарии данные можно записать в файл:

```python
with open("file.txt", "w", encoding="utf-8") as file:
    for result in results:
        file.write(f"{result}\n")
```

Файл подходит, пока программа в основном читает и переписывает данные целиком. По мере роста приложения нужно работать уже не с одним листом данных, а с общей картотекой: находить отдельную запись, менять её, связывать с другими и не давать нескольким операциям испортить общее состояние. Появляются дополнительные требования:

- быстро находить и обновлять отдельные записи;
- согласованно выполнять несколько связанных изменений;
- поддерживать ограничения и связи между данными;
- безопасно обслуживать параллельные запросы;
- управлять восстановлением, резервным копированием и доступом.

СУБД — система управления базами данных — берёт на себя эти механизмы поверх физического хранения. Она не отменяет проектирование данных, но предоставляет согласованный способ хранить, искать, изменять и восстанавливать их. В семинаре рассматриваются основные понятия реляционных баз и интеграция SQLite с Python-приложением.


## Реляционные и нереляционные БД
В реляционной базе данные организованы в таблицы с заранее описанными столбцами. Каждая строка представляет одну запись, первичный ключ однозначно её определяет, а внешний ключ ссылается на запись другой таблицы. Для описания схемы и запросов используется SQL. Ограничения `NOT NULL`, `UNIQUE`, `CHECK` и `FOREIGN KEY` не просто документируют правила, а заставляют базу отклонять недопустимые изменения.

Нереляционные СУБД используют другие модели: ключ—значение, документы, графы или широкие столбцы. Это не означает, что данные лишены структуры. Структура и связи выражаются иначе и часто в большей степени контролируются приложением. Выбор модели зависит от реальных запросов, требований к согласованности, масштаба и характера данных, а не от моды на конкретную СУБД.

Далее рассматриваем реляционную модель и SQLite.


## Транзакции
Транзакция объединяет одну или несколько операций с данными в единое целое. Банковский перевод требует как минимум списать деньги с одного счёта и зачислить на другой. Если выполнить только первое действие, база останется в недопустимом состоянии. Поэтому обе операции подтверждают вместе через `COMMIT`, а при ошибке отменяют через `ROLLBACK`.

Аббревиатура ACID описывает четыре свойства транзакций:

- **Atomicity — атомарность.** Подтверждаются либо все операции транзакции, либо ни одна.
- **Consistency — согласованность.** Успешная транзакция переводит базу из одного допустимого состояния в другое, сохраняя заданные ограничения. Сами бизнес-правила для этого должны быть выражены в схеме и коде корректно.
- **Isolation — изоляция.** Параллельные транзакции не должны видеть недопустимые промежуточные состояния. Конкретные гарантии зависят от уровня изоляции и СУБД.
- **Durability — долговечность.** После успешного `COMMIT` изменения сохраняются при сбое процесса или перезапуске системы в пределах гарантий конкретной СУБД и её настроек.

```sql
BEGIN;
UPDATE accounts SET balance = balance - 100 WHERE id = 1;
UPDATE accounts SET balance = balance + 100 WHERE id = 2;
COMMIT;
```


## SQL, DDL и DML
SQL (Structured Query Language) — язык работы с реляционными базами данных. DDL-команды описывают и изменяют структуру базы: таблицы, ограничения и индексы. DML-команды читают и изменяют сами данные. В примерах ниже используется синтаксис SQLite.

```sql
PRAGMA foreign_keys = ON;

-- DDL: описываем таблицы и ограничения.
CREATE TABLE users (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL UNIQUE,
    password_hash TEXT NOT NULL
);

CREATE TABLE posts (
    id INTEGER PRIMARY KEY,
    user_id INTEGER NOT NULL,
    text TEXT NOT NULL CHECK (length(text) > 0),
    FOREIGN KEY (user_id) REFERENCES users(id)
        ON DELETE CASCADE
        ON UPDATE CASCADE
);

-- DML: добавляем, читаем, обновляем и удаляем данные.
INSERT INTO users (name, password_hash)
VALUES ('user1', '<argon2id hash>');

SELECT id, name FROM users;
SELECT id, name FROM users WHERE name = 'user1';

UPDATE users
SET password_hash = '<new argon2id hash>'
WHERE name = 'user1';

DELETE FROM users WHERE name = 'user1';
```

Пароли нельзя хранить в открытом виде или хешировать быстрыми алгоритмами общего назначения. Для них используют специализированный алгоритм хеширования паролей, например Argon2id.

Значения из пользовательского ввода передают отдельно от SQL-текста. Драйвер сам кодирует параметр как значение, а не как часть команды:

```python
cursor.execute("SELECT id, name FROM users WHERE id = ?", (user_id,))
```

`UPDATE` или `DELETE` без `WHERE` затронет все строки таблицы. Это штатная семантика SQL, а не защита от ошибки разработчика. Перед массовым изменением полезно выполнить `SELECT` с тем же условием и проводить само изменение внутри транзакции.


## NULL
`NULL` — специальный маркер отсутствующего или неизвестного значения. Это не ноль, не пустая строка и не слово `NULL`. SQL использует трёхзначную логику: `TRUE`, `FALSE` и `UNKNOWN`. Поэтому выражения `NULL = NULL`, `value = NULL` и `value != NULL` дают `UNKNOWN`, а `WHERE` оставляет только строки, для которых условие истинно. Для проверки отсутствия значения нужны `IS NULL` и `IS NOT NULL`.

```sql
SELECT * FROM heroes WHERE age IS NULL;
SELECT * FROM heroes WHERE age IS NOT NULL;
SELECT name, COALESCE(age, 0) AS age_or_zero FROM heroes;
```

`COALESCE` возвращает первое значение, отличное от `NULL`. Подмена неизвестного возраста нулём в примере показывает синтаксис, но обычно искажает данные: нулевой возраст и неизвестный возраст означают разное. Значение по умолчанию выбирают по правилам предметной области, а не ради удобства запроса.


## Виды JOIN
`JOIN` формирует результат из пар строк двух таблиц, которые удовлетворяют условию после `ON`. Он не прикрепляет вторую таблицу к первой навсегда, а строит набор строк для конкретного запроса. Пусть в `users` есть Анна, Борис и Вера, у Анны два поста, у Бориса один, а у Веры ни одного.

| Соединение | Какие строки попадут в результат | Что произойдёт с Верой |
|---|---|---|
| `INNER JOIN` | Только строки, для которых нашлась пара | Не попадёт в результат |
| `LEFT JOIN` | Все строки левой таблицы и найденные пары справа | Попадёт один раз, поля поста будут `NULL` |
| `RIGHT JOIN` | Все строки правой таблицы и найденные пары слева | Зависит от порядка таблиц; обычно понятнее переставить таблицы и использовать `LEFT JOIN` |
| `FULL OUTER JOIN` | Все строки обеих таблиц | Попадёт, даже если пары нет |

В связи «один ко многим» одна строка пользователя образует отдельную пару с каждым его постом. Поэтому данные пользователя повторяются в результате запроса: у Анны с двумя постами будет две строки. Это нормальная форма результата `JOIN`, а не дублирование записей в таблице `users`.

```sql
-- Только пользователи, у которых есть посты.
SELECT users.name, posts.text
FROM users
INNER JOIN posts ON posts.user_id = users.id;

-- Все пользователи, включая пользователей без постов.
SELECT users.name, posts.text
FROM users
LEFT JOIN posts ON posts.user_id = users.id;

-- Только пользователи без постов.
SELECT users.id, users.name
FROM users
LEFT JOIN posts ON posts.user_id = users.id
WHERE posts.id IS NULL;
```

Первичный ключ `users.id` не бывает `NULL`, а ограничение `posts.user_id NOT NULL` требует автора для каждого поста. Внешний ключ запрещает ссылаться на несуществующего пользователя. Эти ограничения помещают часть правил предметной области в базу, поэтому их нельзя случайно обойти другим участком приложения.


## Индексы
Индекс — вспомогательная структура данных, которая ускоряет поиск, соединение и сортировку по определённым столбцам. Он выполняет ту же практическую роль, что предметный указатель в книге: вместо последовательного просмотра всех страниц можно быстро перейти к нужным местам. Без подходящего индекса СУБД может просканировать всю таблицу (`full table scan`). Ограничения `PRIMARY KEY` и `UNIQUE` обычно создают необходимые индексы автоматически.

```sql
-- Ускоряем поиск постов пользователя и JOIN по внешнему ключу.
CREATE INDEX IF NOT EXISTS idx_posts_user_id ON posts (user_id);

-- В SQLite можно посмотреть план выполнения запроса.
EXPLAIN QUERY PLAN
SELECT * FROM posts WHERE user_id = 1;
```

Как и отдельный указатель, индекс занимает место и требует обновления при изменении данных. Поэтому индексы ускоряют выбранные операции чтения, но делают `INSERT`, `UPDATE` и `DELETE` дороже. Индексируют столбцы, которые действительно участвуют в важных запросах, а пользу проверяют по плану выполнения и измерениям. Индекс на каждом столбце — не оптимизация.


## ORM

ORM (Object–Relational Mapping) сопоставляет таблицы и связи с Python-классами. Она переводит операции над моделями в SQL, превращает строки результата в объекты и помогает управлять сессией. Но ORM не устраняет саму реляционную модель: за вызовами Python по-прежнему стоят SQL-запросы, ограничения и планы выполнения. Ниже используется стиль SQLAlchemy 2.x.

```python
from __future__ import annotations

from sqlalchemy import ForeignKey, String, create_engine, select
from sqlalchemy.orm import (
    DeclarativeBase,
    Mapped,
    Session,
    mapped_column,
    relationship,
)


class Base(DeclarativeBase):
    pass


class User(Base):
    __tablename__ = "users"

    id: Mapped[int] = mapped_column(primary_key=True)
    name: Mapped[str] = mapped_column(String(255), unique=True, index=True)
    password_hash: Mapped[str] = mapped_column(String(255))
    posts: Mapped[list[Post]] = relationship(
        back_populates="author",
        cascade="all, delete-orphan",
    )


class Post(Base):
    __tablename__ = "posts"

    id: Mapped[int] = mapped_column(primary_key=True)
    user_id: Mapped[int] = mapped_column(
        ForeignKey("users.id", ondelete="CASCADE"),
        index=True,
    )
    text: Mapped[str]
    author: Mapped[User] = relationship(back_populates="posts")


engine = create_engine("sqlite:///database.db")
Base.metadata.create_all(engine)

with Session(engine) as session:
    statement = select(User).where(User.name == "user1")
    user = session.scalars(statement).one_or_none()

    posts_statement = (
        select(Post)
        .join(Post.author)
        .where(User.name == "user1")
    )
    posts = session.scalars(posts_statement).all()
```

`Session` хранит контекст работы ORM с объектами и транзакцией. `session.add(...)` регистрирует объект для сохранения, а `session.commit()` подтверждает накопленные изменения. После ошибки транзакцию откатывают через `session.rollback()`, прежде чем продолжать работу с этой сессией.

Количество SQL-запросов всё равно нужно контролировать. Например, ленивое чтение связи для каждого объекта в списке может выполнить один запрос за списком и ещё по запросу на каждый элемент — это проблема N+1.


## FastAPI, SQLModel, SQLAlchemy и Pydantic: соединяем всё вместе

Табличная модель описывает хранение данных, модель входа — то, что разрешено прислать клиенту, а модель ответа — то, что разрешено вернуть наружу. Поэтому их разделяют. В примере клиент не может назначить серверный `id`, а `secret_name` не попадает в публичный ответ. Таблицы создаются при запуске приложения через `lifespan`.

```python
from contextlib import asynccontextmanager
from typing import Annotated

from fastapi import Depends, FastAPI, HTTPException, Query, Response, status
from sqlmodel import Field, SQLModel, Session, create_engine, select


class HeroBase(SQLModel):
    name: str = Field(index=True, min_length=1)
    age: int | None = Field(default=None, ge=0)


class Hero(HeroBase, table=True):
    id: int | None = Field(default=None, primary_key=True)
    secret_name: str


class HeroCreate(HeroBase):
    secret_name: str


class HeroPublic(HeroBase):
    id: int


class HeroUpdate(SQLModel):
    name: str | None = Field(default=None, min_length=1)
    age: int | None = Field(default=None, ge=0)
    secret_name: str | None = None


sqlite_file_name = "database.db"
sqlite_url = f"sqlite:///{sqlite_file_name}"
connect_args = {"check_same_thread": False}
engine = create_engine(sqlite_url, connect_args=connect_args)


def create_db_and_tables() -> None:
    SQLModel.metadata.create_all(engine)


def get_session():  # одна сессия на запрос
    with Session(engine) as session:
        yield session


SessionDep = Annotated[Session, Depends(get_session)]


@asynccontextmanager
async def lifespan(app: FastAPI):
    create_db_and_tables()
    yield


app = FastAPI(lifespan=lifespan)


@app.post(
    "/heroes/",
    response_model=HeroPublic,
    status_code=status.HTTP_201_CREATED,
)
def create_hero(hero: HeroCreate, session: SessionDep) -> Hero:
    db_hero = Hero.model_validate(hero)
    session.add(db_hero)
    session.commit()
    session.refresh(db_hero)
    return db_hero


@app.get("/heroes/", response_model=list[HeroPublic])
def read_heroes(
    session: SessionDep,
    offset: Annotated[int, Query(ge=0)] = 0,
    limit: Annotated[int, Query(ge=1, le=100)] = 100,
) -> list[Hero]:
    statement = select(Hero).offset(offset).limit(limit)
    return list(session.exec(statement).all())


@app.get("/heroes/{hero_id}", response_model=HeroPublic)
def read_hero(hero_id: int, session: SessionDep) -> Hero:
    hero = session.get(Hero, hero_id)
    if hero is None:
        raise HTTPException(status_code=404, detail="Hero not found")
    return hero


@app.patch("/heroes/{hero_id}", response_model=HeroPublic)
def update_hero(
    hero_id: int, hero: HeroUpdate, session: SessionDep
) -> Hero:
    db_hero = session.get(Hero, hero_id)
    if db_hero is None:
        raise HTTPException(status_code=404, detail="Hero not found")
    update_data = hero.model_dump(exclude_unset=True)
    db_hero.sqlmodel_update(update_data)
    session.add(db_hero)
    session.commit()
    session.refresh(db_hero)
    return db_hero


@app.delete("/heroes/{hero_id}", status_code=status.HTTP_204_NO_CONTENT)
def delete_hero(hero_id: int, session: SessionDep) -> Response:
    hero = session.get(Hero, hero_id)
    if hero is None:
        raise HTTPException(status_code=404, detail="Hero not found")
    session.delete(hero)
    session.commit()
    return Response(status_code=status.HTTP_204_NO_CONTENT)

```

`create_all()` подходит для первого учебного прототипа и создаёт отсутствующие таблицы. Он не является системой изменения существующей схемы. Когда структура рабочего проекта развивается, изменения оформляют миграциями, например через Alembic, а не ручным удалением и пересозданием таблиц.


## Где пригодится в индивидуальном проекте

База данных нужна, если состояние проекта должно сохраняться между запусками: пользователи, настройки, каталог, события или результаты вычислений. До написания таблиц определите, какие сущности существуют в проекте, как они связаны и какие состояния недопустимы. Затем отделите HTTP-модели от табличных.

Запросы к базе лучше сосредоточить в отдельном слое, чтобы обработчики API не отвечали одновременно за разбор HTTP-запроса, SQL и бизнес-правила. Для маленького прототипа этот слой может состоять из нескольких функций; создавать сложную архитектуру заранее не требуется.

Для небольшого локального прототипа обычно подходит SQLite. Если приложение запускается в нескольких экземплярах, получает высокую параллельную нагрузку или требует отдельного управления доступом, резервным копированием и отказоустойчивостью, выбор СУБД нужно пересмотреть по этим требованиям.

## Итоги

- Схема и ограничения фиксируют допустимое состояние данных.
- Транзакция объединяет связанные изменения и позволяет подтвердить или отменить их вместе.
- `JOIN` связывает таблицы, а тип соединения определяет, сохранятся ли строки без пары.
- Индексы ускоряют выбранные операции чтения ценой места и более дорогой записи.
- ORM и SQLModel упрощают интеграцию с Python, но понимание SQL и границ транзакций остаётся необходимым.
