## Общая идея

Если индекс элемента списка известен, доступ занимает `O(1)`. Но для поиска произвольного значения в худшем случае приходится просмотреть весь список — это `O(N)`. Хеш-таблица получает ключ и вычисляет, в какой части хранилища его искать. На этой идее построены `set` и `dict`:

- множество хранит уникальные ключи;
- словарь хранит пары «ключ — значение».

## Путь ключа через хеш-таблицу

```text
ключ ──► hash(ключ) ──► индекс корзины ──► кандидаты ──► сравнение через ==
              │              hash % M              │
              │                                    └── ключ найден / ключа нет
              └── одинаковые ключи обязаны иметь одинаковый хеш
```

В учебной таблице с методом цепочек путь выглядит так. Числа условные:

```text
hash("cat") = 42 ─┐
                     ├── hash % 8 = 2 ──► корзина 2: ["cat", "dog"]
hash("dog") = 18 ─┘                              │
                                                    └── проверить key == candidate

массив корзин:  0: []   1: []   2: [cat, dog]   3: []   ...   7: []
```

Представьте стеллаж с пронумерованными лотками для документов. По фамилии на документе сотрудник вычисляет номер лотка и сразу подходит к нужному месту, но внутри могут лежать документы разных людей. Поэтому после выбора лотка всё равно приходится прочитать фамилию и проверить совпадение.

Хеш-таблица действует так же. Разные ключи могут получить одинаковый хеш, а разные хеши могут превратиться в один индекс корзины. Поэтому одного хеша недостаточно: после выбора корзины таблица сравнивает ключи через `==`. Хеш сокращает число кандидатов, но сам по себе не доказывает равенство.

## Коллизии и коэффициент заполнения

Попадание разных ключей в одну позицию называется **коллизией**. Коллизии неизбежны: возможных ключей гораздо больше, чем ячеек массива. Если документов тысячи, а лотков сто, в некоторых лотках обязательно окажется больше одного документа. Таблица должна уметь хранить и различать такие ключи, иначе данные будут теряться.

Есть два распространённых подхода:

1. **Метод цепочек:** каждая корзина хранит несколько элементов. Именно его нужно реализовать в задании.
2. **Открытая адресация:** элементы лежат в самом массиве, а при занятой позиции алгоритм проверяет другие позиции. `dict` и `set` в CPython используют собственный вариант открытой адресации; это деталь реализации, а не требование языка.

Коэффициент заполнения `α = N / M` показывает отношение числа элементов `N` к числу позиций `M`. Чем больше элементов приходится на доступное место, тем больше кандидатов нужно проверять после вычисления индекса. Номер лотка перестаёт сильно помогать, если лоток набит документами.

Поэтому реализация периодически создаёт массив большего размера и заново распределяет элементы. Такое расширение стоит `O(N)`: нужно пройти по текущим данным и вычислить для них новые позиции. Но расширение происходит не при каждой вставке, поэтому стоимость распределяется по множеству дешёвых операций. Именно это означает амортизированная оценка `O(1)`.

| Операция | Средняя/амортизированная сложность | Худший случай |
|---|---:|---:|
| поиск по ключу | `O(1)` | `O(N)` |
| вставка | `O(1)` | `O(N)` |
| удаление | `O(1)` | `O(N)` |
| обход всех элементов | `O(N)` | `O(N)` |

Худший случай возникает, когда слишком много ключей попадает в одно место или когда вставка запускает расширение таблицы. Поэтому утверждение «операции словаря всегда работают за `O(1)`» неверно. `O(1)` — ожидаемая или амортизированная оценка при равномерном распределении ключей, а не гарантия для каждой отдельной операции.

## Какие объекты можно использовать как ключи

Объект можно использовать как ключ, если его хеш остаётся стабильным на протяжении жизни объекта и сам объект поддерживает сравнение. Главный контракт выглядит так:

```text
a == b  ⇒  hash(a) == hash(b)
```

Обратное утверждение неверно: одинаковый хеш не означает равенство. Если два равных ключа дают разные хеши, таблица отправит их в разные места и не сможет надёжно найти уже сохранённый ключ. Это похоже на ситуацию, когда один и тот же документ при каждом обращении относят в новый лоток.

Числа, строки и `frozenset` обычно хешируемы. У встроенных списков, словарей и множеств хеша нет: их содержимое можно изменить, и хеш на основе содержимого перестал бы быть стабильным. Изменяемость пользовательского объекта сама по себе не является формальным запретом, но состояние, от которого зависят `__eq__` и `__hash__`, нельзя менять, пока объект используется как ключ. Кортеж хешируем только в том случае, если хешируемы все его элементы.

Хеши строк и байтов в Python по умолчанию рандомизируются между запусками процесса. Поэтому значение `hash()` нельзя сохранять в файл, передавать как постоянный идентификатор или использовать вместо криптографического хеша.

## Итоги

- Хеш сокращает множество кандидатов, но равенство ключей подтверждает `==`.
- Коллизии неизбежны и должны разрешаться без потери элементов.
- Средняя работа с ключом занимает `O(1)`, худший случай — `O(N)`.
- Изменение размера удерживает коэффициент заполнения под контролем и даёт амортизированную оценку вставки.
- Хешируемый ключ обязан иметь стабильный хеш; равные ключи обязаны иметь одинаковые хеши.
- Встроенный `dict` сохраняет порядок вставки, но это отдельное свойство, а не общее свойство любой хеш-таблицы.
