### PEP 8: немного об оформлении

[PEP 8](https://peps.python.org/pep-0008/) — рекомендации по оформлению Python-кода. Единообразная запись помогает быстрее замечать смысл и ошибки. Для первых задач достаточно нескольких правил:

- **Четыре пробела на уровень отступа.** Настройте редактор на вставку пробелов; не смешивайте их с табуляцией.
- **Имена переменных и функций — `snake_case`:** `pages_left`, `nickname_status`. Подбирайте имена по смыслу.
- **Пробелы вокруг `=` при присваивании и вокруг сравнений, после запятых.** В аннотациях пишем `name: str` и ` -> bool`; внутри скобок лишние пробелы не нужны.
- **Тело условия, цикла или функции — с новой строки.** Несколько инструкций через `;` затрудняют чтение.
- **В `.py`-файле между определениями функций верхнего уровня — две пустые строки.** Внутри функции пустой строкой можно отделить логический шаг.
- **Ориентир для длины строки — 79 символов.** Длинные выражения можно переносить внутри скобок. В проекте учитывайте его принятые правила оформления.

Отступы определяют блоки кода и влияют на выполнение. Четыре пробела — рекомендация по их ширине; многие нарушения стиля сами по себе не мешают запуску.

## Как выполняется код

В `.py`-файле инструкции выполняются сверху вниз; условия, циклы и вызовы функций меняют порядок. Выражение вычисляет значение, например `120 - 35`; присваивание `pages_left = 120 - 35` связывает имя с результатом. `=` — присваивание, `==` — проверка равенства.

В Jupyter выполненные ячейки используют один процесс Python — **kernel**. Имена сохраняются до его перезапуска. Номер возле ячейки показывает порядок её фактического запуска. **Restart Kernel and Run All Cells** позволяет проверить ноутбук с чистого состояния.

Предскажите, что напечатает пример. Изменится ли `pages_left` само, когда мы переприсвоим `pages_read`?

````python
total_pages = 120
pages_read = 35
pages_left = total_pages - pages_read

pages_read = 40
print(pages_left)

pages_left = total_pages - pages_read
print(pages_left)
````

Сначала выводится `85`, затем `80`. Присваивание сохранило результат вычисления, а не формулу, которая автоматически пересчитывается при изменении других имён.

На протяжении лекции задаём три вопроса: **какое значение получилось, какой у него тип и какая инструкция выполнится следующей?**

## Значения, типы и преобразования

`type(value)` показывает тип, `int("120")` преобразует текст в целое число, `str(120)` — число в текст. Преобразование должно иметь смысл: строку `"книга"` нельзя превратить в целое число через `int`.

Предскажите значения и типы результатов. В Python тип есть у значения; имя можно связать с другим значением.

````python
raw_pages = "120"
total_pages = int(raw_pages)

print(raw_pages + "20")
print(total_pages + 20)
print(type(raw_pages), type(total_pages))
````

## Вычисление и вывод

Обычные `+`, `-`, `*` работают с числами. `/` выполняет деление и возвращает `float`, `//` — деление с округлением вниз, `%` — остаток. Скобки делают порядок вычислений явным.

Книгу из 125 страниц читают по 20 страниц за день. Сколько получится полных порций по 20 страниц и сколько страниц останется?

Строка с префиксом `f` подставляет значения из фигурных скобок. Запись `{value:.2f}` выводит число с двумя знаками после точки; она меняет представление, а не само число.

````python
total_pages = 125
pages_per_day = 20
full_days = total_pages // pages_per_day
remaining_pages = total_pages % pages_per_day

print(f"Полных порций: {full_days}; осталось страниц: {remaining_pages}")
print(f"Частное: {total_pages / pages_per_day:.2f}")
````

## Условия: длина имени в игре

Пусть по правилам игры имя должно содержать **от 3 до 12 символов включительно**. Нам нужны три ответа: `too short`, `too long` и `ok`. Длину строки возвращает `len(name)`.

`if` проверяет первое условие, `elif` — следующее, если предыдущие ложны, `else` выполняется, если ни одно не подошло. После условия ставится двоеточие, тело ветви записывается с отступом в четыре пробела. В одной цепочке выполняется не более одной ветви.

До запуска назовите ответ для `"кот"`. Затем проверьте строки длиной 2, 3, 12 и 13. Почему для проверки недостаточно одного имени средней длины?

````python
name = "кот"
length = len(name)

if length < 3:
    status = "too short"
elif length > 12:
    status = "too long"
else:
    status = "ok"

print(status)
````

## Несколько условий и truthiness

Для ответа «подходит или нет» достаточно `3 <= len(name) <= 12`. Это та же проверка, что `len(name) >= 3 and len(name) <= 12`.

`and` требует истинности обоих условий, `or` — хотя бы одного, `not` меняет логическое значение на противоположное. В логическом контексте ложны `False`, `None`, числовой ноль, пустые строки и контейнеры; большинство остальных значений истинны. `bool(value)` позволяет явно увидеть это значение.

`if name:` проверяет непустоту, но не допустимую длину. `if pages_read is None:` проверяет отсутствие данных: **0 прочитанных страниц и неизвестное число страниц — разные случаи**.

````python
name = "я"
print(bool(name))
print(3 <= len(name) <= 12)

pages_read = 0
if pages_read is None:
    print("Число прочитанных страниц неизвестно")
else:
    print(f"Прочитано страниц: {pages_read}")
````

**Короткое замыкание:** `and` и `or` не вычисляют правую часть, если левая уже определила исход проверки. Объясните, почему следующий пример не делит на ноль. Потом замените `days` на `3`.

````python
days = 0
pages_read = 60
fast_reading = days > 0 and pages_read / days >= 20
print(fast_reading)
````

## От проверки к функции

Чтобы проверять разные имена, вынесем правило в функцию. `def` создаёт функцию; её тело выполняется при вызове. `name` в определении — **параметр**, строка `"кот"` при вызове — **аргумент**.

`return` немедленно завершает функцию и передаёт результат вызывающему коду. Если выполнение дошло до конца тела без `return`, результатом будет `None`.

`name: str` и `-> str` — аннотации ожидаемых типов аргумента и результата. Они помогают читать код, но сам интерпретатор не проверяет соответствие им автоматически.

Перед запуском найдите строку, на которой закончится каждый вызов.

````python
def nickname_status(name: str) -> str:
    length = len(name)
    if length < 3:
        return "too short"
    if length > 12:
        return "too long"
    return "ok"


print(nickname_status("я"))
print(nickname_status("кот"))
print(nickname_status("оченьдлинноеимя"))
````

## Проверяем обычный случай и границы

Функция получает данные через параметры и возвращает результат. Теперь её можно проверить без ручной замены имён в программе.

`assert condition` ничего не выводит, если условие истинно, и сообщает об `AssertionError`, если оно ложно. Начнём с обычного случая, затем проверим пустую строку и значения с обеих сторон границ.

Запись `"a" * 12` повторяет строку 12 раз. Какой `assert` обнаружит ошибку, если в функции заменить `< 3` на `<= 3`?

````python
assert nickname_status("игрок") == "ok"
assert nickname_status("") == "too short"
assert nickname_status("aa") == "too short"
assert nickname_status("aaa") == "ok"
assert nickname_status("a" * 12) == "ok"
assert nickname_status("a" * 13) == "too long"
````

## Почему `print` не заменяет `return`

`print` выводит значение для человека. `return` передаёт результат программе, которая вызвала функцию. Предскажите весь вывод и значение переменной `result`.

````python
def show_length(text: str):
    print(len(text))


result = show_length("кот")
print(result)
````

Первый вызов напечатал `3`, а второй — `None`: в `show_length` нет `return`. Чтобы использовать длину в дальнейших вычислениях, функция должна возвращать `len(text)`.

Параметры и имена, созданные внутри функции, локальны. Например, `length` внутри `nickname_status` не меняет одноимённую переменную снаружи. Проверим это отдельным вызовом.

````python
length = 100
result = nickname_status("кот")
print(result, length)
````

## `for`: читаем заданное число дней

За первый день прочитали 10 страниц, за второй — 20, за третий — 30. Сколько всего прочитано? Перед запуском проследите значения `day` и `pages_read` по шагам.

`for` перебирает значения. `range(1, 4)` даёт `1`, `2`, `3`: правая граница не включается. Перед циклом создаём накопитель, внутри обновляем его. `pages_read += today` означает `pages_read = pages_read + today`.

````python
pages_read = 0

for day in range(1, 4):
    today = day * 10
    pages_read += today
    print(day, today, pages_read)

print(f"Всего страниц: {pages_read}")
````

## `while`: читаем до конца книги

Теперь книга содержит 65 страниц, а за день можно прочитать до 20. Нужно определить число дней; последний день может быть неполным.

`while` проверяет условие перед каждой итерацией. В теле должно меняться состояние, от которого зависит продолжение цикла. Здесь будем уменьшать число оставшихся страниц, не опуская его ниже нуля: `min(a, b)` возвращает меньшее из двух значений.

До запуска назовите ответ. Затем проверьте книги из 0, 20 и 21 страницы. Почему при 0 страниц тело цикла не выполняется ни разу?

````python
pages_left = 65
pages_per_day = 20
days = 0

while pages_left > 0:
    today = min(pages_left, pages_per_day)
    pages_left -= today
    days += 1

print(days)
````

## `break` и `continue`

`break` завершает ближайший цикл. `continue` пропускает оставшуюся часть текущей итерации и переходит к следующей.

В следующем примере пропустим дни, кратные трём, и остановим чтение, когда книга закончится. На остальных днях читаем до 20 страниц. Операция `day % 3 == 0` проверяет кратность трём.

Предскажите, какие дни будут напечатаны. Пальцем проследите переходы: куда уводит `continue`, а куда — `break`?

````python
pages_left = 65

for day in range(1, 8):
    if day % 3 == 0:
        continue
    pages_left -= min(pages_left, 20)
    print(day, pages_left)
    if pages_left == 0:
        break
````

## Как прочитать ошибку

**Намеренно ошибочный пример.** Он показан текстом, чтобы общий запуск ноутбука не останавливался. Объясните причину, затем запустите исправленный вариант в следующей ячейке.

```python
pages = "120"

```

При запуске такого `.py`-файла конец сообщения выглядит так:

```text
  File "example.py", line 2, in <module>

```

Читаем снизу вверх: **тип ошибки** `TypeError`, **причина** — попытка соединить строку с числом, **место** — строка 2. В Jupyter вместо имени файла будет ссылка на ячейку. Если в сообщении несколько вызовов, начинайте с последнего участка вашего кода.

Исправление зависит от задачи: для числового сложения нужно `int(pages) + 20`, для соединения текста — `pages + "20"`.

````python
pages = "120"
result = int(pages) + 20
assert result == 140
print(result)
````

## Имена, выражения и операции

- Python различает регистр: `name` и `Name` — разные имена. Имена могут содержать буквы, цифры и `_`, но не начинаются с цифры и не совпадают с ключевыми словами.
- `#` начинает комментарий. Число `10_000` равно `10000`: подчёркивание помогает читать разряды.
- `+`, `-`, `*`, `/`, `//`, `%`, `**` — сложение, вычитание, умножение, деление, деление с округлением вниз, остаток и степень.
- `==`, `!=`, `<`, `<=`, `>`, `>=` — сравнения. В сложном выражении порядок лучше обозначать скобками.
- `range(stop)` начинает с 0, `range(start, stop)` — с `start`; третий аргумент задаёт шаг. Правая граница всегда исключена. Например, `range(0, 7, 2)` даёт 0, 2, 4, 6.
- `input()` возвращает строку. В задачах этого занятия данные приходят через параметры, поэтому `input()` не нужен; подробнее разберём ввод на занятии 2.

## Именованные аргументы и значения по умолчанию

Аргументы можно передать по позиции или по имени параметра. Значение по умолчанию используется, если аргумент не передан.

Сигнатура описывает параметры функции; аннотации дополнительно указывают ожидаемые типы параметров и результата. Следующая функция по умолчанию проверяет длину от 3, а при явном аргументе `minimum` — от указанной границы.

````python
def is_long_enough(text: str, minimum: int = 3) -> bool:
    return len(text) >= minimum


print(is_long_enough("кот"))
print(is_long_enough("кот", 5))
print(is_long_enough(minimum=2, text="ёж"))
````

# Дополнительно: изменяемое значение по умолчанию

Списки подробно появятся на занятии 2, общие ссылки и изменяемость — на занятии 5.

Для чтения примера достаточно трёх пояснений: `[]` создаёт список, `append` добавляет элемент в существующий список, `a is b` проверяет, один ли это объект.

Предскажите содержимое `first` и `second` после двух вызовов. Мы печатаем оба результата только после второго вызова.

````python
def add_note(note: str, notes=[]):
    notes.append(note)
    return notes


first = add_note("прочитать главу")
second = add_note("решить задачу")
print(first)
print(second)
print(first is second)
````

**Разбор после запуска.** Значение по умолчанию создаётся один раз при выполнении `def`. Оба вызова без второго аргумента получают один список, поэтому оба имени показывают две заметки, а `first is second` даёт `True`.

Чтобы получать новый список на каждый вызов, используют `None` и создают список внутри функции. Правило относится к изменяемым значениям по умолчанию; передача аргумента по имени сама по себе его не меняет.

````python
def add_note_safely(note: str, notes=None):
    if notes is None:
        notes = []
    notes.append(note)
    return notes


safe_first = add_note_safely("прочитать главу")
safe_second = add_note_safely("решить задачу")
print(safe_first)
print(safe_second)
print(safe_first is safe_second)
````
