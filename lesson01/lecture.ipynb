## Модель выполнения

Интерпретатор читает исходный код, разбирает его и исполняет инструкции. В обычном `.py`-файле выполнение начинается с первой строки и идёт сверху вниз. Условия, циклы и вызовы функций меняют этот порядок.

В Jupyter каждая выполненная ячейка изменяет состояние одного процесса Python — **kernel**. Имена из выполненной ячейки доступны следующим ячейкам, пока kernel не перезапущен. Номер слева показывает порядок фактического запуска, поэтому ячейка ниже может быть выполнена раньше ячейки выше. Для проверки воспроизводимости полезно периодически выполнять **Restart Kernel and Run All Cells**.

````python
balance = 1_000
balance = balance + 250
balance -= 100
print(balance)  # Предскажите результат до выполнения ячейки.
````

## Арифметические операции

- `+`, `-`, `*` — сложение, вычитание и умножение;
- `/` — обычное деление, результат имеет тип `float`;
- `//` — деление с округлением вниз;
- `%` — остаток от деления;
- `**` — возведение в степень.

Сначала выполняются скобки, затем степень, умножение и деление, затем сложение и вычитание. Когда порядок важен читателю, лучше поставить явные скобки.

````python
price = 1_250
discount_percent = 12
final_price = price * (1 - discount_percent / 100)
print(final_price)

print(17 // 5, 17 % 5)
print(2 ** 10)
````

## Truthiness

Условию не обязательно передавать именно `True` или `False`: Python приводит значение к логическому контексту. Ложными считаются `False`, `None`, числовой ноль и пустые строки или контейнеры. Большинство остальных значений истинны — в том числе строка `"0"`.

Поэтому `if text:` обычно означает «строка непустая», а `if value is None:` — «значение отсутствует». Это разные проверки: ноль может быть корректным наблюдением, а `None` — отсутствием наблюдения.

````python
observations = 0
comment = ""
growth_rate = None

print(bool(observations), bool(comment), bool(growth_rate))
print(bool("0"), bool(-1), bool(0.01))

if not comment:
    print("Комментарий не заполнен")
if growth_rate is None:
    print("Темп роста не рассчитан")
````

## Сравнения

Сравнение возвращает `True` или `False`:

- `==`, `!=` — равно и не равно;
- `<`, `<=`, `>`, `>=` — сравнение порядка.

Не путайте присваивание `=` и проверку равенства `==`. Сравнения можно объединять: `21 <= age <= 65`.

````python
age = 30
income = 90_000
print(age >= 21)
print(21 <= age <= 65)
print(income == 90_000)
print(income != 0)
````

## Логические операции

- `a and b` истинно, когда истинны оба условия;
- `a or b` истинно, когда истинно хотя бы одно;
- `not a` меняет логическое значение на противоположное.

Операции вычисляются **лениво**: если левый операнд уже определил результат, правый не вычисляется. Это называют short-circuit evaluation. Сложное условие полезно разбивать на именованные части — тогда программу легче проверить и объяснить.

````python
age = 30
income = 90_000
has_overdue = False

age_ok = 21 <= age <= 65
income_ok = income >= 50_000
approved = age_ok and income_ok and not has_overdue
print(approved)

count = 0
total = 100
has_positive_average = count > 0 and total / count > 0
print(has_positive_average)  # Деление на ноль не выполнялось.
````

## Ветвление `if`

`if` выполняет блок, когда условие истинно. `elif` проверяет следующее условие, только если предыдущие ложны. `else` выполняется, когда ни одно условие не подошло.

После условия ставится двоеточие. Тело ветви записывается с одинаковым отступом, обычно четыре пробела. В одной цепочке выполняется не более одной ветви.

````python
revenue = 500
expenses = 450

if revenue > expenses:
    result = "profit"
elif revenue < expenses:
    result = "loss"
else:
    result = "break-even"

print(result)
````

## Порядок условий

Условия проверяются сверху вниз, поэтому более узкий случай часто должен стоять раньше широкого. Для интервальных правил удобно использовать цепочку `if/elif/else`, а для одновременных требований — одно выражение с `and`.

Перед написанием кода полезно выписать границы и проверить значения непосредственно слева, на границе и справа от неё.

````python
income = 1_500_000

if income <= 1_000_000:
    rate = 0.10
elif income <= 3_000_000:
    rate = 0.15
else:
    rate = 0.20

print(rate)
# Это выбор единой ставки, а не расчёт налога по отдельным частям дохода.
````

## Цикл `for` и `range`

`for` в Python перебирает элементы итерируемого объекта, а не является специальной формой счётчика. Для числового диапазона используют `range(start, stop, step)`; правая граница `stop` не включается. Тело цикла задаётся отступом.

````python
total = 0

for number in range(1, 6):
    total += number

print(total)
print(list(range(0, 10, 2)))
````

## Цикл `while`

`while` повторяет блок, пока условие истинно. Он подходит, когда число повторений заранее неизвестно. В теле должно изменяться состояние, от которого зависит условие, иначе цикл не завершится.

````python
capital = 100.0
target = 130.0
rate = 0.10
periods = 0

while capital < target:
    capital *= 1 + rate
    periods += 1

print(periods, f"{capital:.2f}")
````

## `break` и `continue`

`break` немедленно завершает ближайший цикл. `continue` пропускает остаток текущей итерации. Эти операторы полезны для раннего выхода и фильтрации, но большое количество переходов обычно ухудшает читаемость.

````python
first_divisor = None
number = 91

for candidate in range(2, number):
    if number % candidate != 0:
        continue
    first_divisor = candidate
    break

print(first_divisor)
````

## Функции и `return`

Инструкция `def` создаёт функцию, но не выполняет её тело. Тело запускается при вызове. Параметры в определении получают значения аргументов вызова, а `return` немедленно завершает функцию и возвращает результат. Если `return` не встретился, функция возвращает `None`.

`print` показывает значение человеку, но не передаёт его вызывающему коду — поэтому не заменяет `return`. Аннотации `name: type` и `-> type` документируют ожидаемые типы, но сам Python обычно не проверяет их во время выполнения.

````python
def financial_result(revenue: int, expenses: int) -> str:
    """Классифицировать финансовый результат."""
    if revenue > expenses:
        return "profit"
    if revenue < expenses:
        return "loss"
    return "break-even"


print(financial_result(500, 450))
print(financial_result(500, 500))
````

## Параметры и аргументы

В вызове аргументы можно передать по позиции или явно указать имя параметра. Параметру можно задать значение по умолчанию — тогда соответствующий аргумент необязателен. Именованные аргументы особенно полезны, когда несколько значений одного типа легко перепутать.

Сигнатура функции — её имя, параметры и аннотация возвращаемого типа. Например, `format_amount(amount: float, currency: str = "RUB") -> str`.

````python
def format_amount(amount: float, currency: str = "RUB") -> str:
    return f"{amount:.2f} {currency}"


print(format_amount(1250))
print(format_amount(1250, "USD"))
print(format_amount(currency="EUR", amount=99.9))
````

## Области видимости

Имена, созданные внутри функции, относятся к её **локальной области видимости** и снаружи недоступны. Параметры тоже локальны. Имя, созданное вне функции, находится во внешней области, но скрытая зависимость от него усложняет проверку кода.

Практическое правило первого занятия: необходимые данные передавайте параметрами, промежуточные значения храните в локальных именах, результат возвращайте через `return`.

````python
rate = 0.10

def calculate_tax(income: float, rate: float) -> float:
    tax = income * rate  # rate и tax здесь локальны.
    return tax


print(calculate_tax(200_000, 0.13))
print(rate)
# print(tax)  # NameError: локальное имя tax снаружи функции недоступно.
````

## Функция как контракт

У функции должны быть понятные входы, результат и поведение на границах. Несколько `assert` превращают примеры в исполняемую проверку: если условие ложно, Python сообщит об `AssertionError`. Полноценное тестирование появится позже, но привычка проверять обычные и граничные случаи нужна уже сейчас.

````python
assert financial_result(500, 450) == "profit"
assert financial_result(450, 500) == "loss"
assert financial_result(500, 500) == "break-even"
````

## Как читать простую ошибку

Последняя строка traceback сообщает тип и причину ошибки, строки выше — место возникновения. Частые ошибки первого занятия:

- `SyntaxError` — нарушен синтаксис, например пропущено двоеточие;
- `NameError` — имя ещё не создано или написано иначе;
- `TypeError` — операция не поддерживает такие типы;
- `ValueError` — тип преобразования подходит, но содержимое строки имеет неверный формат.

Не исправляйте ошибку наугад: сначала прочитайте её тип, строку и значения участвующих переменных.

Попробуйте по очереди выполнить выражения в отдельной пустой ячейке и прочитать последнюю строку сообщения:

```python
"100" + 20       # TypeError: складываются несовместимые типы
int("12.5")       # ValueError: строка не записана как целое число
unknown_name + 1   # NameError: имя не определено
```

После наблюдения удалите ошибочную ячейку или перезапустите kernel. Явная обработка ошибок появится на занятии 2.

## Неожиданно, но по правилам

Перед запуском предскажите результат каждого выражения. Здесь нет специальных исключений из изученных правил — непривычен только их результат.

1. Непустая строка истинна независимо от написанного внутри текста. Поэтому строки `"False"` и `"0"` не являются ложными значениями.
2. `//` округляет вниз, а не просто отбрасывает дробную часть. Для отрицательного числа «вниз» означает в сторону меньшего числа. Остаток `%` согласован с равенством `a == (a // b) * b + a % b`.
3. Аннотация типа описывает контракт, но интерпретатор не запрещает передать значение другого типа. Оператор `*` затем работает по правилам фактического объекта.

````python
print(bool("False"), bool("0"))

quotient = -7 // 2
remainder = -7 % 2
print(quotient, remainder)
print(quotient * 2 + remainder)

def double(value: int) -> int:
    return value * 2

print(double(3))
print(double("ha"))  # Аннотация int не остановила вызов.
````
